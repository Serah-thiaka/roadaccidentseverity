```python
import pandas as pd

df = pd.read_csv('sample_data/ecommerce_customer_churn_dataset.csv')
print(df.head(3))
```

```python
# Select our numerical customer behaviors
features = ['Age', 'Session_Duration_Avg', 'Cart_Abandonment_Rate']

X = df[features]
y = df['Churned']

print("Features selected successfully! Here is a preview of your training columns:")
print(X.head(2))
```

```python
# Fill missing data values with the column average (mean)
X = X.fillna(X.mean())

# Check to ensure there are zero missing numbers left
print("Missing values remaining in features:")
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

# Split the cleaned data into training pools (80%) and testing pools (20%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print(f"Data split successfully!")
print(f"Training rows: {X_train.shape[0]} | Testing rows: {X_test.shape[0]}")
```

```python
from sklearn.ensemble import RandomForestClassifier

# 1. Initialize the Random Forest Classifier model
model = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)

# 2. Train the model using your 40,000 training rows
model.fit(X_train, y_train)

# 3. Calculate and print the final accuracy score on your test rows
accuracy = model.score(X_test, y_test)
print(f" Random Forest Model Trained Successfully!")
print(f"Your Customer Behaviour Model Accuracy is: {accuracy:.2%}")
```

```python
# Extract the importance score of each metric
importances = model.feature_importances_
features_list = ['Age', 'Session_Duration_Avg', 'Cart_Abandonment_Rate']

print("---  Customer Behaviour Importance Weights ---")
for feature, importance in zip(features_list, importances):
    print(f"Customer Metric: {feature:<25} | Weight: {importance:.2%}")
```

```python
from sklearn.metrics import confusion_matrix

# 1. Generate the missing predictions from our test features
predictions = model.predict(X_test)

# 2. Now calculate the breakdown of right vs wrong answers
cm = confusion_matrix(y_test, predictions)

print("--- Model Error Breakdown Matrix ---")
print(f"Correctly Predicted as 'Stayed': {cm[0][0]}")
print(f"Incorrectly Predicted as 'Left' (False Alarm): {cm[0][1]}")
print(f"Incorrectly Predicted as 'Stayed' (Missed Risk): {cm[1][0]}")
print(f"Correctly Predicted as 'Left': {cm[1][1]}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Gather your feature list and their exact calculated values from the model
features_list = ['Age', 'Session_Duration_Avg', 'Cart_Abandonment_Rate']
importances = model.feature_importances_

# 2. Sort them so the highest importance bar sits nicely at the top
indices = np.argsort(importances)

# 3. Plot the horizontal bar chart
plt.figure(figsize=(9, 5))
plt.title('E-commerce Customer Behaviour: Feature Importance Profile', fontsize=14, fontweight='bold', pad=15)
bars = plt.barh(range(len(indices)), importances[indices], color=['#4C72B0', '#55A868', '#C44E52'], edgecolor='black', height=0.6)

# 4. Add formatting and labels
plt.yticks(range(len(indices)), [features_list[i] for i in indices], fontsize=11)
plt.xlabel('Relative Predictive Weight (Percentage)', fontsize=12, labelpad=10)
plt.xlim(0, 0.65) # Give some padding on the right side for the labels

# 5. Add exact percentage labels onto the bars themselves
for bar in bars:
    width = bar.get_width()
    plt.text(width + 0.01, bar.get_y() + bar.get_height()/2, f'{width:.2%}',
             va='center', ha='left', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```
